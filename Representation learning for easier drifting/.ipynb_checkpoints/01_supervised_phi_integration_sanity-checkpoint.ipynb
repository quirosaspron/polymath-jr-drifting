# Supervised phi integration sanity check

This notebook tests the engineering path for a learned representation before we design the unsupervised drift-aware objective. A small `PhiNetwork` is trained to reproduce the toy problem's known two-dimensional signal, frozen, inserted into the DriftXpress kernel, and compared with raw and oracle representations.

This is **not** the final scientific method: because the training target is the known signal, this experiment only asks whether a neural representation can learn, freeze, cache, and drive Xpress correctly.

```python
from pathlib import Path
import time
import matplotlib.pyplot as plt
import torch

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = (PROJECT_ROOT / "..").resolve()
if not (PROJECT_ROOT / "src").exists():
    raise RuntimeError(f"Could not find src/ from {Path.cwd()}")
import os
os.chdir(PROJECT_ROOT)

from src.data import make_paired_run_inputs, make_toy_drift_problem
from src.diagnostics import plot_discrepancy_curves
from src.evaluators import make_toy_evaluator
from src.experiments import ExperimentConfig, run_comparison, save_comparison, seed_everything
from src.models import PhiNetwork
from src.representation import freeze_representation, train_phi_to_known_signal, validate_representation
from src.toy import run_particle_evolution
```

## Hypothesis and fixed protocol

The immediate hypothesis is: **a neural `phi` trained to filter the known nuisance coordinates should approach the known-signal Xpress result when it is frozen and used inside the kernel.**

The drift comparison keeps the target, initial particles, landmarks, temperature, step size, checkpoints, and seeds fixed. The three methods differ only in the representation passed to the Xpress field.

```python
DRIFT_SEEDS = (7, 17, 27, 37, 47, 57)
TOY_KWARGS = dict(
    n_samples=512,
    ambient_dim=32,
    signal_dim=2,
    centers=8,
    cluster_std=0.6,
    nuisance_scale=1.0,
    initial_signal_scale=1.5,
    batch_size=256,
    n_landmarks=128,
    temperature=1.0,
    step_size=0.1,
    backend="xpress",
)
PHI_KWARGS = dict(
    latent_dim=TOY_KWARGS["ambient_dim"],
    feature_dim=TOY_KWARGS["signal_dim"],
    hidden_dims=(64, 64),
    residual=False,
)
PHI_STEPS = 500
PHI_LEARNING_RATE = 1e-3

DRIFT_CONFIG = ExperimentConfig(
    dataset="toy32",
    methods=("raw", "known_signal", "learned_phi"),
    seeds=DRIFT_SEEDS,
    drift_steps=100,
    eval_every=5,
    primary_metric="signal_swd",
    fixed_budget=100,
    output_dir=Path("artifacts/toy_supervised_phi_sanity"),
)
DRIFT_CONFIG, PHI_KWARGS
```

## Train phi to the known signal

Each seed gets its own toy target and its own phi initialization. The helper below uses ordinary mean-squared error only for this integration check. After training, `freeze_representation` disables parameter gradients so the production Xpress cache can be built.

```python
phi_by_seed = {}
phi_history_by_seed = {}
phi_health_by_seed = {}
phi_fit_by_seed = {}
phi_training_seconds = {}

for seed in DRIFT_SEEDS:
    seed_everything(seed)
    problem = make_toy_drift_problem(
        n_samples=TOY_KWARGS["n_samples"],
        ambient_dim=TOY_KWARGS["ambient_dim"],
        signal_dim=TOY_KWARGS["signal_dim"],
        centers=TOY_KWARGS["centers"],
        cluster_std=TOY_KWARGS["cluster_std"],
        nuisance_scale=TOY_KWARGS["nuisance_scale"],
        initial_signal_scale=TOY_KWARGS["initial_signal_scale"],
        random_state=seed,
    )
    phi = PhiNetwork(**PHI_KWARGS)
    started = time.perf_counter()
    history = train_phi_to_known_signal(
        phi=phi,
        inputs=problem.target,
        target_features=problem.target_signal,
        steps=PHI_STEPS,
        batch_size=None,
        learning_rate=PHI_LEARNING_RATE,
        seed=seed + 1000,
    )
    phi_training_seconds[seed] = time.perf_counter() - started
    freeze_representation(phi)
    phi_by_seed[seed] = phi
    phi_history_by_seed[seed] = history
    with torch.no_grad():
        phi_fit_by_seed[seed] = float(torch.nn.functional.mse_loss(phi(problem.target), problem.target_signal))
    phi_health_by_seed[seed] = validate_representation(phi=phi, fixed_latents=problem.target)

print("phi training seconds:", {seed: round(value, 2) for seed, value in phi_training_seconds.items()})
print("final signal-recovery MSE:", {seed: round(value, 5) for seed, value in phi_fit_by_seed.items()})
print("one health report:", phi_health_by_seed[DRIFT_SEEDS[0]])
```

```python
figure, axes = plt.subplots(1, 2, figsize=(12, 4))
for seed, history in phi_history_by_seed.items():
    axes[0].plot(history.steps, history.losses, label=str(seed))
axes[0].set_title("Known-signal recovery loss")
axes[0].set_xlabel("optimization step")
axes[0].set_ylabel("MSE(phi(x), signal(x))")
axes[0].legend(title="seed")

health_values = [phi_health_by_seed[seed] for seed in DRIFT_SEEDS]
axes[1].bar(
    [str(seed) for seed in DRIFT_SEEDS],
    [report.effective_rank for report in health_values],
)
axes[1].axhline(2.0, color="black", linestyle="--", label="two signal dims")
axes[1].set_title("Frozen phi effective rank")
axes[1].set_xlabel("seed")
axes[1].set_ylabel("effective rank")
axes[1].legend()
figure.tight_layout()
display(figure)
```

## Freeze phi and run DriftXpress

This is the important integration step. The learned representation is now treated exactly like the oracle representation: it is frozen, its target-side Xpress quantities are cached once per run, and only the moving particles are updated.

```python
def paired_factory(seed):
    return make_paired_run_inputs(
        n_steps=DRIFT_CONFIG.drift_steps,
        dataset_size=TOY_KWARGS["n_samples"],
        batch_size=TOY_KWARGS["batch_size"],
        noise_dim=1,
        n_landmarks=TOY_KWARGS["n_landmarks"],
        evaluation_size=1,
        seed=seed,
    )

def run_one(run_config, method, seed, paired_inputs):
    problem = make_toy_drift_problem(
        n_samples=TOY_KWARGS["n_samples"],
        ambient_dim=TOY_KWARGS["ambient_dim"],
        signal_dim=TOY_KWARGS["signal_dim"],
        centers=TOY_KWARGS["centers"],
        cluster_std=TOY_KWARGS["cluster_std"],
        nuisance_scale=TOY_KWARGS["nuisance_scale"],
        initial_signal_scale=TOY_KWARGS["initial_signal_scale"],
        random_state=seed,
    )
    representation = {
        "raw": None,
        "known_signal": problem.project_signal,
        "learned_phi": phi_by_seed[seed],
    }[method]
    return run_particle_evolution(
        problem=problem,
        paired_inputs=paired_inputs,
        representation=representation,
        evaluator=make_toy_evaluator(problem),
        method=method,
        seed=seed,
        temperature=TOY_KWARGS["temperature"],
        step_size=TOY_KWARGS["step_size"],
        eval_every=DRIFT_CONFIG.eval_every,
        backend=TOY_KWARGS["backend"],
        output_dir=DRIFT_CONFIG.output_dir / method / f"seed_{seed}",
    )

comparison = run_comparison(
    DRIFT_CONFIG,
    run_one=run_one,
    paired_factory=paired_factory,
)
for row in comparison.summary:
    print(
        f"{row['method']}: final={row['final_mean']:.4f} +/- {row['final_std']:.4f}; "
        f"AUC={row['auc_mean']:.4f} +/- {row['auc_std']:.4f}"
    )
```

```python
for metric_name in ("signal_swd", "observed_swd", "nuisance_swd"):
    figure, axis = plot_discrepancy_curves(
        comparison,
        metric_name=metric_name,
        show_individual_seeds=True,
    )
    display(figure)

comparison_path = save_comparison(comparison)
comparison_path
```

## Interpretation

The first question is whether the recovery loss decreases and the frozen representation has two non-collapsed features. The second is whether `learned_phi` approaches `known_signal` on the external Xpress curves.

Even if this works, it is only an engineering sanity check because phi was trained with the true signal. The exploratory section below replaces `train_phi_to_known_signal` with a user-authored-style one-step drift-aware objective while keeping the frozen-drift comparison unchanged.

## Exploratory drift-aware objective (one-step progress)

The supervised section above used the forbidden shortcut `target_signal`. This section removes that shortcut. It trains phi only from observed target samples and initial probe particles.

For each phi update: (1) split target samples into training, probe-loss, and held-out evaluation sets; (2) build a differentiable Xpress cache from the training target; (3) compute an Xpress field for fixed probe particles; (4) take one hypothetical particle step; and (5) minimize a differentiable MMD proxy between the moved probes and the probe-loss target set.

The cache is rebuilt after every phi update because changing phi changes all target-side kernel quantities. After phi is frozen, the ordinary cached Xpress path is used again.

```python
from src.evaluators import DistributionEvaluator
from src.field import build_differentiable_xpress_cache, compute_xpress_field

def torch_mmd_rbf(x, y, bandwidth):
    def kernel(left, right):
        squared_distances = torch.cdist(left.float(), right.float()).square()
        return torch.exp(-squared_distances / (2.0 * bandwidth ** 2))

    return (
        kernel(x, x).mean()
        + kernel(y, y).mean()
        - 2.0 * kernel(x, y).mean()
    )

def median_bandwidth(values):
    distances = torch.pdist(values.float())
    positive = distances[distances > 0]
    return float(positive.median()) if positive.numel() else 1.0

def train_drift_aware_phi(
    problem,
    *,
    seed,
    steps=100,
    probe_size=128,
    train_size=256,
    loss_size=128,
    n_landmarks=64,
    learning_rate=1e-3,
    temperature=1.0,
    step_size=0.1,
):
    if train_size + loss_size >= problem.target.shape[0]:
        raise ValueError("the held-out evaluation split must be non-empty")
    if probe_size > problem.initial.shape[0]:
        raise ValueError("probe_size exceeds the initial particle count")

    split_generator = torch.Generator(device="cpu").manual_seed(seed + 4000)
    permutation = torch.randperm(problem.target.shape[0], generator=split_generator)
    train_indices = permutation[:train_size]
    loss_indices = permutation[train_size : train_size + loss_size]
    eval_indices = permutation[train_size + loss_size :]
    target_train = problem.target[train_indices].detach()
    target_for_loss = problem.target[loss_indices].detach()
    target_for_eval = problem.target[eval_indices].detach()
    landmark_indices = torch.randperm(train_size, generator=split_generator)[:n_landmarks]
    probe_particles = problem.initial[:probe_size].detach()
    bandwidth = median_bandwidth(target_for_loss)

    seed_everything(seed + 3000)
    phi = PhiNetwork(**PHI_KWARGS)
    optimizer = torch.optim.Adam(phi.parameters(), lr=learning_rate)
    history = []
    phi.train()
    for step in range(steps):
        differentiable_cache = build_differentiable_xpress_cache(
            target_train,
            landmark_indices=landmark_indices,
            temperature=temperature,
            representation=phi,
        )
        field = compute_xpress_field(
            probe_particles,
            cache=differentiable_cache,
            representation=phi,
        )
        moved_probe = probe_particles + step_size * field
        loss = torch_mmd_rbf(moved_probe, target_for_loss, bandwidth)
        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(phi.parameters(), max_norm=10.0)
        optimizer.step()
        history.append(
            {
                "step": step + 1,
                "loss": float(loss.detach().cpu()),
                "field_norm": float(field.detach().norm(dim=1).mean().cpu()),
            }
        )
    freeze_representation(phi)
    return phi, history, target_for_eval
```

```python
DRIFT_AWARE_SEEDS = (7, 17, 27)  # pilot only; expand after inspecting stability
DRIFT_AWARE_STEPS = 100

drift_aware_phi_by_seed = {}
drift_aware_history_by_seed = {}
drift_aware_eval_target_by_seed = {}
drift_aware_health_by_seed = {}

for seed in DRIFT_AWARE_SEEDS:
    problem = make_toy_drift_problem(
        n_samples=TOY_KWARGS["n_samples"],
        ambient_dim=TOY_KWARGS["ambient_dim"],
        signal_dim=TOY_KWARGS["signal_dim"],
        centers=TOY_KWARGS["centers"],
        cluster_std=TOY_KWARGS["cluster_std"],
        nuisance_scale=TOY_KWARGS["nuisance_scale"],
        initial_signal_scale=TOY_KWARGS["initial_signal_scale"],
        random_state=seed,
    )
    phi, history, eval_target = train_drift_aware_phi(
        problem,
        seed=seed,
        steps=DRIFT_AWARE_STEPS,
        temperature=TOY_KWARGS["temperature"],
        step_size=TOY_KWARGS["step_size"],
    )
    drift_aware_phi_by_seed[seed] = phi
    drift_aware_history_by_seed[seed] = history
    drift_aware_eval_target_by_seed[seed] = eval_target
    drift_aware_health_by_seed[seed] = validate_representation(
        phi=phi, fixed_latents=problem.target
    )

print("final one-step proxy loss:", {seed: round(history[-1]["loss"], 5) for seed, history in drift_aware_history_by_seed.items()})
print("frozen phi health:", drift_aware_health_by_seed)
```

```python
figure, axes = plt.subplots(1, 2, figsize=(12, 4))
for seed, history in drift_aware_history_by_seed.items():
    axes[0].plot(
        [item["step"] for item in history],
        [item["loss"] for item in history],
        label=str(seed),
    )
axes[0].set_title("Drift-aware one-step proxy loss")
axes[0].set_xlabel("phi optimization step")
axes[0].set_ylabel("MMD after one probe update")
axes[0].legend(title="seed")

for seed, history in drift_aware_history_by_seed.items():
    axes[1].plot(
        [item["step"] for item in history],
        [item["field_norm"] for item in history],
        label=str(seed),
    )
axes[1].set_title("Probe field magnitude")
axes[1].set_xlabel("phi optimization step")
axes[1].set_ylabel("mean field norm")
axes[1].legend(title="seed")
figure.tight_layout()
display(figure)
```

## Freeze the drift-aware phi and compare external Xpress curves

The final comparison uses a held-out target split that was not used in the one-step proxy loss. The raw, known-signal, and drift-aware methods share each seed's initial particles and paired Xpress inputs.

```python
def heldout_toy_evaluator(problem, target_reference):
    views = {
        "observed": lambda values: values,
        "signal": problem.project_signal,
    }
    if problem.nuisance_projection.shape[1] > 0:
        views["nuisance"] = problem.project_nuisance
    return DistributionEvaluator(target_reference, views=views)

DRIFT_AWARE_CONFIG = ExperimentConfig(
    dataset="toy32",
    methods=("raw", "known_signal", "drift_aware_phi"),
    seeds=DRIFT_AWARE_SEEDS,
    drift_steps=100,
    eval_every=5,
    primary_metric="signal_swd",
    fixed_budget=100,
    output_dir=Path("artifacts/toy_drift_aware_phi_probe"),
)

def drift_aware_paired_factory(seed):
    return make_paired_run_inputs(
        n_steps=DRIFT_AWARE_CONFIG.drift_steps,
        dataset_size=TOY_KWARGS["n_samples"],
        batch_size=TOY_KWARGS["batch_size"],
        noise_dim=1,
        n_landmarks=TOY_KWARGS["n_landmarks"],
        evaluation_size=1,
        seed=seed,
    )

def drift_aware_run_one(run_config, method, seed, paired_inputs):
    problem = make_toy_drift_problem(
        n_samples=TOY_KWARGS["n_samples"],
        ambient_dim=TOY_KWARGS["ambient_dim"],
        signal_dim=TOY_KWARGS["signal_dim"],
        centers=TOY_KWARGS["centers"],
        cluster_std=TOY_KWARGS["cluster_std"],
        nuisance_scale=TOY_KWARGS["nuisance_scale"],
        initial_signal_scale=TOY_KWARGS["initial_signal_scale"],
        random_state=seed,
    )
    representation = {
        "raw": None,
        "known_signal": problem.project_signal,
        "drift_aware_phi": drift_aware_phi_by_seed[seed],
    }[method]
    evaluator = heldout_toy_evaluator(
        problem, drift_aware_eval_target_by_seed[seed]
    )
    return run_particle_evolution(
        problem=problem,
        paired_inputs=paired_inputs,
        representation=representation,
        evaluator=evaluator,
        method=method,
        seed=seed,
        temperature=TOY_KWARGS["temperature"],
        step_size=TOY_KWARGS["step_size"],
        eval_every=DRIFT_AWARE_CONFIG.eval_every,
        backend=TOY_KWARGS["backend"],
        output_dir=DRIFT_AWARE_CONFIG.output_dir / method / f"seed_{seed}",
    )

drift_aware_comparison = run_comparison(
    DRIFT_AWARE_CONFIG,
    run_one=drift_aware_run_one,
    paired_factory=drift_aware_paired_factory,
)
for row in drift_aware_comparison.summary:
    print(
        f"{row['method']}: final={row['final_mean']:.4f} +/- {row['final_std']:.4f}; "
        f"AUC={row['auc_mean']:.4f} +/- {row['auc_std']:.4f}"
    )
```

```python
for metric_name in ("signal_swd", "observed_swd", "nuisance_swd"):
    figure, axis = plot_discrepancy_curves(
        drift_aware_comparison,
        metric_name=metric_name,
        show_individual_seeds=True,
    )
    display(figure)

drift_aware_comparison_path = save_comparison(drift_aware_comparison)
drift_aware_comparison_path
```

## What this pilot shows
We decressed the loss function but we have higher signal swd than raw, so the objective is misaligned.
